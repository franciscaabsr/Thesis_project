```python
# Script for creating graphs from fMRI and EEG connectivity data (coverting MATLAB matrices)

import scipy.io
import networkx as nx 
import bct 
import numpy as np

#1.Creation of average Graph
def createAvGraph(file):
    
    # Load Matlab file with connectivity matrix for each time point - 3D matrix
    mat = scipy.io.loadmat(file)

    conn_fMRI = mat['connFMRI'].transpose() #connectivity matrix with numpy format
    print(conn_fMRI.shape[0])
    t_points = conn_fMRI.shape[0] # number of layers of multilayer matrix
    num_areas = conn_fMRI.shape[1] #number of nodes of the graph
    
    # Get average connectivity matrix
    av_conn = np.zeros((68,68))
       
    for i in range(t_points):
        
        av_conn = av_conn + conn_fMRI[i]
        #sprint(av_conn)
            
    av_conn = av_conn/t_points 
    print(av_conn)
    
    G = nx.from_numpy_matrix(av_conn)
    
    return G

#2. Creation of array of graphs (equivalent to layers)
def createArrayGraph(file):
    
    # Load Matlab file with connectivity matrix for each time point - 3D matrix
    mat = scipy.io.loadmat(file)

    conn_fMRI = mat['connFMRI'].transpose() #connectivity matrix with numpy format
    print(conn_fMRI.shape[0])
    t_points = conn_fMRI.shape[0] # number of layers of multilayer matrix
    num_areas = conn_fMRI.shape[1] #number of nodes of the graph
    
    array_graphs = np.empty(t_points, dtype=object) 
    
    for i in range(t_points):
        array_graphs[i] = nx.from_numpy_matrix(conn_fMRI[i])
        
    return array_graphs
    
    
G_av = createAvGraph('/home/francisca/Documents/Tese/Implementation/Datasets/source_reconstructed_FC/fmri_connect_desikan/subj01-7T/conn_desi_phase_coh_time_fmri.mat')
G_array = createArrayGraph('/home/francisca/Documents/Tese/Implementation/Datasets/source_reconstructed_FC/fmri_connect_desikan/subj01-7T/conn_desi_phase_coh_time_fmri.mat')
```
